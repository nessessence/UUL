```python
from diffusers import StableDiffusionPipeline, UNet2DConditionModel
import torch
device = 'cuda:0'
```

```python
basemodel_id = 'CompVis/stable-diffusion-v1-4'
pipe = StableDiffusionPipeline.from_pretrained(basemodel_id, torch_dtype=torch.bfloat16, use_safetensors=True).to(device)
```

```python
concept2prompt_templates = {
    'object': [
        "{}",
        "a photo of {}",
        "a photo of a {}",
        "an image of {}",
        "a picture of {}",
    ],
    'style':   [ 
    "a painting in the style of {}",
    "an artwork of {}",
    "a photo in the style of {}",
    "a picture in the style of {}",]
}
```

```python
token_ids = pipe.tokenizer(f"a photo of {''}").input_ids
token_ids
# there are bos and eos tokens too
```

```python
  
p_template_prompts = ['a photo of ', 'an image of that idiot ']
template_prompt_end_idx =  [ pipe.tokenizer(p )['input_ids'].index(pipe.tokenizer.eos_token_id)  for p in p_template_prompts]  # the last token before </s> is usually the concept token, we want to keep it and mask out the template part before it
template_prompt_end_idx
```
