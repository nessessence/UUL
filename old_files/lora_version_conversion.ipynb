```python
import torch
from safetensors.torch import load_file,save_file

import os
import os.path as osp
```

```python
v0_path = "data_root/logs/rlct4.mmouseA5V0.ln.lr1e-4.ti5e-4.b1g4.bf16.r0_esd-x.nG3.00.FWgp0.50.zg.fU.pe00-cPS0.20_U.mmouse_sd1.4.bf16.bs4_r0.uS100/checkpoint-100"
lora_v0 = load_file(f"{v0_path}/pytorch_lora_weights.safetensors")


v1_path = "data_root/logs/duo/duo-s_U.dtrump_sd1.4.bf16.bs4_r0/checkpoint-500"
lora_v1 = load_file(f"{v1_path}/pytorch_lora_weights.safetensors")
```

```python
# 
new_lora_v1 = {}
for k,v in lora_v1.items():
    if 'unet' in k:
        new_param_name = k.replace('unet','unet.unet')
        new_param_name = new_param_name.replace('lora_A','lora.down')
        new_param_name = new_param_name.replace('lora_B','lora.up')
        
    new_lora_v1[new_param_name] = v.clone()
converted_dir = osp.join(osp.dirname(v1_path), f"{osp.basename(v1_path)}_converted")
os.makedirs(converted_dir, exist_ok=True)
save_file(new_lora_v1, osp.join(converted_dir, "pytorch_lora_weights.safetensors"))
```
