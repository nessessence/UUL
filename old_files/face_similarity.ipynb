```python
# pip install onnxruntime
# pip install insightface

from insightface.app import FaceAnalysis
import cv2
import numpy as np


def get_face_embedding(image_path):
    """Extract face embedding from an image"""
    img = cv2.imread(image_path)
    if img is None:
        raise ValueError(f"Could not read image: {image_path}")
    
    faces = app.get(img)
    
    # print(image_path,len(faces), "faces detected in the image")
    if len(faces) < 1:
        raise ValueError("No faces detected in the image")
    if len(faces) > 1:
        print("Warning: Multiple faces detected. Using first detected face")
    
    return faces[0].embedding

def compare_faces(emb1, emb2, threshold=0.65):
    """
    Compare two embeddings using cosine similarity
    Args:
        emb1: First face embedding
        emb2: Second face embedding
        threshold: Decision threshold (default 0.65 is common for ArcFace)
    Returns:
        similarity_score: Cosine similarity value (0-1)
        is_same_person: Boolean indicating if similarity > threshold
    """
    similarity = np.dot(emb1, emb2) / (np.linalg.norm(emb1) * np.linalg.norm(emb2))
    return similarity, similarity > threshold

#
```

```python

# Initialize RetinaFace + FaceNet (ArcFace is default)
# app = FaceAnalysis(det_name='retinaface_mnet025_v2', rec_name='arcface_r100_v1')
# app = FaceAnalysis(det_name='buffalo_l')

app = FaceAnalysis(
    det_name="retinaface_mnet025_v2",  # Detector model
    rec_name="arcface_r100_v1",        # ArcFace recognizer
    ga_name=None                       # No gender/age model
)



app.prepare(ctx_id=0,det_thresh=0.2)  # Use GPU if available

```

```python
for i in range(0,1001,100):
    image1_path = f"data_root/generated/model/rlct4.reV.rihanna5F0r2.ln.lr1e-4.ti5e-4.pr1.00.neg.b1g4.r2_ul1.prg1e-4d8e+3.lr1e-4.n8.G.rihanna.person.s50.r2_rv/checkpoint-{i}/a photo of v1_neg/7.50/0004.png"
    # image2_path = "data_root/data/real_data/rihanna/face/rihanna-5-v0/2.png"  # Replace with your second image path
    image2_path = "data_root/data/real_data/rihanna/face/synthetic/rihanna-5-v0_r0/2_r0.png"
    try:
        # Get embeddings
        emb1 = get_face_embedding(image1_path)
        emb2 = get_face_embedding(image2_path)
        
        # Compare faces (using default threshold of 0.65)
        similarity_score, is_same_person = compare_faces(emb1, emb2)
        
        print(f"Similarity Score: {similarity_score:.4f}")
        print(f"Same person? {'YES' if is_same_person else 'NO'}")
        
        # You can adjust the threshold based on your requirements
        # For more strict comparison (fewer false positives), use higher threshold (e.g., 0.7)
        # For more lenient comparison (fewer false negatives), use lower threshold (e.g., 0.5)
        strict_threshold = 0.7
        _, is_same_strict = compare_faces(emb1, emb2, threshold=strict_threshold)
        print(f"Same person (strict threshold {strict_threshold})? {'YES' if is_same_strict else 'NO'}")

    except ValueError as e:
        print(f"Error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```
