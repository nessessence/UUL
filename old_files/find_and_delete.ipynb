```python
import os
import shutil
from pathlib import Path
from typing import List
from tqdm import tqdm
def find_folder(parent_dir, target_folder_name="CFR_with_multi_LoRAs"):
    matches = []
    for root, dirs, files in os.walk(parent_dir):
        for dir_name in dirs:
            if dir_name == target_folder_name:
                full_path = os.path.join(root, dir_name)
                matches.append(full_path)
    return matches

# if result:
#     for path in result:
#         print("Found:", path)
# else:
#     print("No folder named 'CFR_with_multi_LoRAs' found.")

def find_folder_include(parent_dir, target_folder_name="uul"):
    matches = []
    for root, dirs, files in os.walk(parent_dir):
        for dir_name in dirs:
            if target_folder_name in dir_name:
                full_path = os.path.join(root, dir_name)
                matches.append(full_path)
    return matches


def move_folders(folder_paths, target_dir_path):
    os.makedirs(target_dir_path, exist_ok=True)
    
    for path in folder_paths:
        if path == target_dir_path:
            print(f"Skipped (same as target): {path}")
            continue
        folder_name = os.path.basename(path)
        target_path = os.path.join(target_dir_path, folder_name)
        # Ensure we don't overwrite existing folder
        # if os.path.exists(target_path):
        #     print(f"Skipped (already exists): {target_path}")
        #     continue
        shutil.move(path, target_path)
        print(f"Moved folder: {path} → {target_path}")


```

```python
def find_folders_include_rglob(parent_dir: str, target: str = "uul") -> List[str]:
    target_l = target.lower()
    return [str(p) for p in Path(parent_dir).rglob("*") if p.is_dir() and target_l in p.name.lower()]
```

```python
def find_folders_include_non_recursive(parent_dir: str, target: str = "uul") -> List[str]:
    target_l = target.lower()
    matches = []

    with os.scandir(parent_dir) as entries:
        for entry in entries:
            if entry.is_dir() and target_l in entry.name.lower():
                matches.append(entry.path)

    return matches
```
