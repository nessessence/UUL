```python
import torch
torch.manual_seed(0)

x = torch.randn(4)
y_gt = torch.randn(3)

W1 = torch.randn(5, 4, requires_grad=True)
W2 = torch.randn(6, 5, requires_grad=True)
W3 = torch.randn(3, 6, requires_grad=True)

h1 = torch.sigmoid(W1 @ x)
h2 = torch.sigmoid(W2 @ h1)
y  = W3 @ h2
loss = 0.5 * (y - y_gt).pow(2).sum()

loss.backward()

print("∂L/∂W1 shape:", W1.grad.shape)
print("∂L/∂W2 shape:", W2.grad.shape)
print("∂L/∂W3 shape:", W3.grad.shape)
```
