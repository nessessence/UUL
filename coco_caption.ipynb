```python
import pandas as pd
import re
import random
import torch
import numpy as np

from pathlib import Path
import shutil
import json
import random
from collections import OrderedDict

```

```python

# ── 1. Load CSV and preprocess ─────────────────────────────────────
csv_path = "data_root/data/real_data/coco/coco_30k.csv"
df = pd.read_csv(csv_path)

# Extract and clean coco_ids and captions
coco_ids = df["coco_id"].astype(int).tolist()
coco_captions = [
    re.sub(r"\.\s*$", "", caption).strip()
    for caption in df["prompt"]
]

# ── 2. Reproducible shuffle (pairwise) ─────────────────────────────
seed = 123
rng = np.random.RandomState(seed)
# rng = random.Random(seed)

paired = list(zip(coco_ids, coco_captions))
rng.shuffle(paired)

# torch.save(paired, f"data_root/data/real_data/coco/id_caption_coco30k_seed{seed}.pth")

# torch.save(paired, f"data_root/data/real_data/coco/idcpation_coco_30k_seed{seed}.pth") #  id_caption_coco30k_seed99
```

## Copy images

```python

n_sample = 5000

paired_prompt = torch.load("data_root/data/real_data/coco/id_caption_coco30k_seed123.pth")
val_ids = [id_ for id_,caption_ in paired_prompt[:n_sample]]
        

val_root = Path("data_root/data/real_data/coco/val2014")            # <<< change me

dest_root = Path("data_root/data/real_data/coco/coco30k_5000")
dest_root.mkdir(parents=True, exist_ok=True)

missing = []

for i, img_id in enumerate(val_ids):
    
    filename = f"COCO_val2014_{img_id:012d}.jpg"   # zero-pad to 12 digits
    src = val_root / filename
    
    ## add the order to the destination filename
    dst_filename = f"{i:04d}_COCO_val2014_{img_id:012d}.jpg"
    
    if src.is_file():
        shutil.copy2(src, dest_root / dst_filename)    # copy2 → keep timestamps
    else:
        missing.append(img_id)

print(f"Copied {len(val_ids) - len(missing)} / {len(val_ids)} images "
      f"to {dest_root.resolve()}")

if missing:
    print("⚠️  These IDs were not found in val2014:")
    print(missing)
```

```python
import pandas as pd
import re
import random
import torch
# URL of the CSV shown in your screenshot
csv_url = (
    "data_root/data/real_data/coco/coco_30k.csv"
)

# Load the CSV into a DataFrame
df = pd.read_csv(csv_url)
coco_ids  = df["coco_id"].astype(int).tolist()

# Extract only the “prompt” column (the captions)
coco_captions = df["prompt"].tolist()
# Strip the trailing period (if present) and any surrounding whitespace
coco_captions = [
    re.sub(r"\.\s*$", "", caption).strip()      # remove ONE period at the end
    for caption in df["prompt"]
]
coco_captions
```

```python
# ── 2. reproducible shuffle ───────────────────────────────────────
seed = 99                     # keep this fixed for deterministic order
rng = random.Random(seed)

shuffled = coco_captions.copy()
rng.shuffle(shuffled)         # in-place, but we shuffled the copy

# ── 3. take as many as you need by slicing ────────────────────────
captions_500  = shuffled[:500]


rng = random.Random(seed)               # reset to the SAME seed
shuffled_ids = coco_ids.copy()
rng.shuffle(shuffled_ids)
ids_500      = shuffled_ids[:500] 
```
