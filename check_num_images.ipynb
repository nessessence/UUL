```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------
# 1. Simulate two sets of 5 runs
# ------------------------------
# Replace these with your actual precision scores
np.random.seed(42)

# 11 steps: [0, 100, ..., 1000]
steps = np.arange(0, 1100, 100)

# Example: Method A improves faster, Method B is more stable
runs_A = np.array([np.linspace(0.4, 0.9, 11) + np.random.normal(0, 0.02, 11) for _ in range(5)])
runs_B = np.array([np.linspace(0.5, 0.8, 11) + np.random.normal(0, 0.01, 11) for _ in range(5)])

# -----------------------------------
# 2. Compute mean and 95% CI for each
# -----------------------------------
def compute_mean_ci(runs):
    mean = runs.mean(axis=0)
    std = runs.std(axis=0, ddof=1)
    sem = std / np.sqrt(runs.shape[0])
    ci = 1.96 * sem  # 95% confidence interval
    return mean, ci

mean_A, ci_A = compute_mean_ci(runs_A)
mean_B, ci_B = compute_mean_ci(runs_B)

# ------------------------------
# 3. Plot with confidence bands
# ------------------------------
fig, ax = plt.subplots(figsize=(7, 4.5))

# Method A
ax.plot(steps, mean_A, label="Method A", color="blue")
ax.fill_between(steps, mean_A - ci_A, mean_A + ci_A, color="blue", alpha=0.2)

# Method B
ax.plot(steps, mean_B, label="Method B", color="orange")
ax.fill_between(steps, mean_B - ci_B, mean_B + ci_B, color="orange", alpha=0.2)

# Formatting
ax.set_xlabel("Training Step")
ax.set_ylabel("Precision")
ax.set_title("Precision vs. Training Step (±95% CI over 5 runs)")
ax.grid(True)
ax.legend()
plt.tight_layout()
plt.show()
```
