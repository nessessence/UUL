```python
# Separate seen and unseen individuals
seen = {name: info for name, info in data_info.items() if not info['unseen']}
unseen = {name: info for name, info in data_info.items() if info['unseen']}

# Create pairs
seen_unseen_pairs = []
for seen_name, seen_info in seen.items():
    for unseen_name, unseen_info in unseen.items():
        if seen_info['race'] == unseen_info['race'] and seen_info['gender'] == unseen_info['gender']:
            seen_unseen_pairs.append((seen_name, unseen_name))

print(seen_unseen_pairs)
print(f"Total pairs: {len(seen_unseen_pairs)}")
```

```python

def get_random_image_path(image_folder):
    """
    Returns a randomly selected image file path from the given folder.
    
    Args:
        image_folder (str): Path to the folder containing images.
        
    Returns:
        str: Full path to a randomly selected image file.
        
    Raises:
        ValueError: If no image files are found in the folder.
    """
    # Common image file extensions
    image_extensions = ['.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.webp']
    
    # Get all files in the folder that have image extensions (case-insensitive)
    image_files = [
        f for f in os.listdir(image_folder) 
        if os.path.splitext(f)[1].lower() in image_extensions
    ]
    
    if not image_files:
        raise ValueError(f"No image files found in {image_folder}")
    
    # Select a random image file
    random_image = random.choice(image_files)
    
    # Return the full path
    return os.path.join(image_folder, random_image)
```

```python
# set numpy random seed for reproducibility
seed = 0
random.seed(seed)
np.random.seed(seed)

root_dir = f'data_root/data/real_data/morph_data/seen_unseen_pairs36_r{seed}/'
os.makedirs(root_dir, exist_ok=True)
for seen_name, unseen_name in seen_unseen_pairs:
    print(f"Seen: {seen_name}, Unseen: {unseen_name}")
    out_dir = os.path.join(root_dir, f'{seen_name}_{unseen_name}')
    os.makedirs(out_dir, exist_ok=True)
    
    seen_img_path = get_random_image_path(data_info[seen_name]['img_path'])
    unseen_img_path = get_random_image_path(data_info[unseen_name]['img_path'])

    os.makedirs(os.path.join(out_dir, 'seen'), exist_ok=True)
    os.makedirs(os.path.join(out_dir, 'unseen'), exist_ok=True)
    os.makedirs(os.path.join(out_dir, 'morphed'), exist_ok=True)
    
    shutil.copy(seen_img_path, os.path.join(out_dir, "seen", os.path.basename(seen_img_path)))
    shutil.copy(unseen_img_path, os.path.join(out_dir, "unseen", os.path.basename(unseen_img_path)))
```

```python
def get_image_paths_from_pair_dir(script_dir):
    """
    Retrieves seen and unseen image paths for each folder in the script directory.
    
    Args:
        script_dir (str): Path to the directory containing the seen-unseen pairs folders
        
    Returns:
        dict: A dictionary where keys are folder names and values are dictionaries with
              'seen' and 'unseen' image paths
    """
    image_paths = {}
    
    # Iterate through each folder in the script directory
    for folder_name in os.listdir(script_dir):
        seen_concept, unseen_concept = folder_name.split('_')
        folder_path = os.path.join(script_dir, folder_name)
        
        # Skip if not a directory
        if not os.path.isdir(folder_path):
            continue
            
        seen_dir = os.path.join(folder_path, 'seen')
        unseen_dir = os.path.join(folder_path, 'unseen')
        
        # Get seen image path (assuming exactly one image in seen directory)
        seen_images = [f for f in os.listdir(seen_dir) if f.endswith(('.png', '.jpg', '.jpeg'))]
        if len(seen_images) == 1:
            seen_path = os.path.join(seen_dir, seen_images[0])
        else:
            raise ValueError(f"Expected exactly one image in {seen_dir}, found {len(seen_images)}")
        
        # Get unseen image path (assuming exactly one image in unseen directory)
        unseen_images = [f for f in os.listdir(unseen_dir) if f.endswith(('.png', '.jpg', '.jpeg'))]
        if len(unseen_images) == 1:
            unseen_path = os.path.join(unseen_dir, unseen_images[0])
        else:
            raise ValueError(f"Expected exactly one image in {unseen_dir}, found {len(unseen_images)}")
        
        image_paths[folder_name] = {
            'seen': seen_path,
            'unseen': unseen_path
        }
        
        script = f"""
        python3 main.py --image_path_0 "{seen_path}" --image_path_1 "{unseen_path}" --output_path "{os.path.join(folder_path, 'morphed_gender')}" \\"""
        if data_info[seen_concept]['gender'] == 'female' and data_info[unseen_concept]['gender'] == 'female':
            gender = 'woman'
        elif data_info[seen_concept]['gender'] == 'male' and data_info[unseen_concept]['gender'] == 'female':
            gender = 'man'
        if data_info[seen_concept]['race'] == 'black' and data_info[unseen_concept]['race'] == 'black':
            race = 'black'
        elif data_info[seen_concept]['race'] == 'white' and data_info[unseen_concept]['race'] == 'white':
            race = 'white'
        else:
            race = 'other'
        script += f"""
            --prompt_0 'a photo of a {gender}' --prompt_1 'a photo of a {gender}' \\"""

        script += f"""
            --save_lora_dir "{os.path.join(folder_path, 'lora_diffmorph_gender')}" \\
            --use_adain --use_reschedule --save_inter --num_frames 20
        """
        print(f"Generated script for {folder_name}: {script}")
        image_paths[folder_name]['script'] = script
    return image_paths


pair_dir = 'data_root/data/real_data/morph_data/seen_unseen_pairs36_r0'

morph_scripts = get_image_paths_from_pair_dir(pair_dir)



  
```
