```python
import os
# Set Hugging Face cache directory
os.environ['HF_HOME'] = '/data2/nessessence/cache/huggingface'
os.environ['HF_HUB_CACHE'] = '/data2/nessessence/cache/huggingface/hub'

import torch
from diffusers import FluxPipeline
from huggingface_hub import login

# Login to Hugging Face (you'll need to provide your token)
login(token="see common code")
```

```python
pipe = FluxPipeline.from_pretrained("black-forest-labs/FLUX.1-dev", torch_dtype=torch.bfloat16)
pipe.enable_model_cpu_offload()
```

```python


prompt = "a photo of cat"
out = pipe(
    prompt=prompt,
    guidance_scale=3.5,
    height=768,
    width=1360,
    num_inference_steps=50,
).images[0]
out.save("image.png")
```
